```python
import os, uuid, hashlib, platform, getpass

# ===== 1. 填写学号 =====
STUDENT_ID = 20244080130

# ===== 2. 打印 conda 环境 =====
CONDA_ENV = os.environ.get("CONDA_DEFAULT_ENV", "base")
print(f"[Conda 环境] {CONDA_ENV}")

# =====3.机器指纹=====
machine_info = f"{platform.node()}|{platform.machine()}|{platform.system()}|{uuid.getnode()}"
MACHINE_FP = hashlib.sha256(machine_info.encode()).hexdigest()[:16]

# =====4.生成唯一ID=====
RUN_ID = f"{STUDENT_ID}-{MACHINE_FP}"
print(f"[学号 {STUDENT_ID}]")
print(f"[机器指纹] {MACHINE_FP}")
print(f"[RUN_ID] {RUN_ID}")
```

```python
# 1. say_hi：无参数，打印 Hello, world!
def say_hi():
    print("Hello, world!")

# 2. goat_pad：接收字符串，前后拼接goat打印
def goat_pad(s):
    print("goat" + s + "goat")

#3. print_n：字符串s打印n次，各占一行
def print_n(s, n):
    for _ in range(n):
        print(s)

# 可以测试调用
say_hi()
goat_pad("test")
print_n("hi",3)
```

```python
def gcd(m,n):
    while n != 0:
        m, n = n, m % n
    return m

# 测试三组数字
print(gcd(2018,2019))
print(gcd(1200,300))
print(gcd(5040,60))
```

```python
import math

#1. euler_limit 极限近似，x=n
def euler_limit(n):
    return (1 + 1/n)**n


#2. euler_infinite_sum：前n项和，n是项数；0! = 1
def euler_infinite_sum(n):
    if n == 0:
        return 0.0
    total = 0.0
    fact = 1   #存阶乘，初始0! = 1
    for k in range(n):
        if k>0:
            fact = fact * k
        total += 1.0 / fact
    return total

#测试例子 euler_infinite_sum(4) → k=0,1,2,3四项：1+1+1/2+1/6 ≈2.66666
print(euler_infinite_sum(4))


#3. euler_approx(epsilon)：循环累加直到和真实e误差小于ϵ
def euler_approx(eps):
    real_e = math.exp(1)
    k = 0
    total = 0.0
    fact = 1
    while abs(total - real_e) >= eps:
        if k>0:
            fact *= k
        total += 1.0 / fact
        k +=1
    return total


#4. print_euler_sum_table(n)：打印k=1~n每一项euler_infinite_sum(k)
def print_euler_sum_table(n):
    for k in range(1, n+1):
        val = euler_infinite_sum(k)
        print(f"k={k}, sum={val}")

#测试
print("真实e", math.exp(1))
print(euler_approx(0.01))
print_euler_sum_table(5)
```

```python
def is_power_of_2(n):
    # n是正整数
    while n > 1:
        if n % 2 != 0:
            return False
        n = n // 2
    return True


def is_power(b, n):
    # 判断 n 是否等于 b^p，p≥0，b,n正整数
    if n ==1:
        return True # b^0 = 1
    while n % b == 0:
        n = n // b
    return n == 1

#测试
print(is_power_of_2(8)) #True
print(is_power_of_2(7)) #False

print(is_power(3, 81)) #True 3^4=81
print(is_power(2,10)) #False
```
